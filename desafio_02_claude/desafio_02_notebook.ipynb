```python
import pandas as pd
import numpy as np
import json
```

```python
with open('churn_nordeste_raw.json', 'r') as file:
    json_bruto = json.load(file)
```

```python
json_bruto
```

```python
dados_normalizados = pd.json_normalize(json_bruto, record_path=['clientes'])
```

```python
dados_normalizados
```

```python
dados_normalizados.info()
```

```python
dados_normalizados[dados_normalizados.select_dtypes('str').columns] = dados_normalizados.select_dtypes('str').apply(lambda x: x.str.strip())
```

```python
contagem = (dados_normalizados == '').sum()
```

```python
total_vazios = contagem[contagem > 0].sum()
```

```python
colunas = contagem[contagem > 0]
```

```python
print(f'{total_vazios} células foram convertidas de string vazia/espaço nas colunas: ')
for x,y in colunas.items():
    print(x)
```

```python
dados_normalizados[dados_normalizados.select_dtypes('str').columns] = dados_normalizados.select_dtypes('str').apply(lambda x: x.replace('', np.nan))
```

```python
dados_normalizados
```

```python
dados_normalizados
```

```python
dados_normalizados['conta.cobranca_mensal'] = dados_normalizados['conta.cobranca_mensal'].apply(lambda x: x.replace(',', '.'))
```

```python
dados_normalizados['conta.cobranca_total'] = dados_normalizados['conta.cobranca_total'].apply(lambda x: x.replace(',', '.'))
```

```python
dados_normalizados.info()
```

```python
dados_normalizados['conta.cobranca_mensal'] = pd.to_numeric(dados_normalizados['conta.cobranca_mensal'], errors='coerce')
```

```python
dados_normalizados['conta.cobranca_total'] = pd.to_numeric(dados_normalizados['conta.cobranca_total'], errors='coerce')
```

```python
dados_normalizados.info()
```
